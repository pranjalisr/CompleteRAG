### Text File Loader

```python
from langchain_community.document_loaders import TextLoader

loader = TextLoader("../data/sample.txt")  #loading the text file
data = loader.load()  #loading the data from the text file
```

```python
len(data)  #checking the length of the data
```

```python
data[0].metadata  #checking the metadata of the data
```

```python
print(data[0].page_content)  #checking the page content of the data
```

#### CSV Loader

```python
from langchain_community.document_loaders import CSVLoader
```

```python
loader = CSVLoader(file_path="../data/sample.csv", encoding="utf-8")  #loading the csv file
```

```python
data = loader.load()  #loading the data from the csv file
```

```python
len(data)  #checking the length of the data
```

```python
print(data[0].page_content)  #checking the page content of the data
```
