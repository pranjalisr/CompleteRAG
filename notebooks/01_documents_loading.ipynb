### Document Loading

```python
from langchain_core.documents import Document
```

```python
doc = Document(    #Document always has 2 objects with page content and metadata
    page_content="Hello, The complete course fees is 23000 Rs.",
    metadata={"page": 1, 
              "book-name": "user.pdf",  
              "author": "Pranjali Srivastava",  
              "total_pages": 9,
              "user_id":"1" },
)

# metadata helps to define details of the data
```
